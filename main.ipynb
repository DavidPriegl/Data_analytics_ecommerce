```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
df['Category'].value_counts(dropna=False)
```

```python
df['Category'] = df['Category'].str.strip().str.title()
```

```python
df['Category'].value_counts(dropna=False)
```

```python
df['CustomerType'] = df['CustomerType'].str.strip().str.title()
```

```python
df['Region'].value_counts(dropna=False)
```

```python
df['Region'] = df['Region'].str.strip().str.title()
```

```python
df['Region'].value_counts(dropna=False)
```

```python
df['UnitPrice'].head(20)
```

```python
df['UnitPrice'] = (
    df['UnitPrice']
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
)

df['UnitPrice'] = pd.to_numeric(df['UnitPrice'], errors='coerce')
```

```python
df['UnitPrice'].dtype
```

```python
df['OrderDate'] = pd.to_datetime(
    df['OrderDate'], format='mixed', errors='coerce'
)
```

```python
df['OrderDate'].head()
```

```python
df['OrderDate'].isnull().sum()
```

```python
df['Category'] = df['Category'].fillna('Unknown')
df['Region'] = df['Region'].fillna('Unknown')
df['CustomerType'] = df['CustomerType'].fillna('Unknown')
df['ShippingMethod'] = df['ShippingMethod'].fillna('Unknown')
```

```python
df['GrossRevenue'] = df['Quantity'] * df['UnitPrice']
```

```python
df['Revenue'] = (
    df['Quantity'] * df['UnitPrice'] * (1-df['Discount'])
)
```

```python
df['Revenue'] = df['Revenue'].round(2)
```

```python
df[
    ['Quantity', 'UnitPrice', 'Discount', 'Revenue', 'CustomerRating']
].describe()
```

```python
total_revenue = df['Revenue'].sum()

print(f'Total Revenue: ${total_revenue:,.2f}')
```

```python
total_orders = df['OrderID'].nunique()

print(f'Total Order: {total_orders:,}')
```

```python
total_customers = df['CustomerID'].nunique()

print(f'Total Customers: {total_customers:,}')
```

```python
average_order_value = total_revenue / total_orders

print(f'Average Order Value: ${average_order_value:,.2f}')
```

```python
revenue_by_product = (
    df.groupby('Product')['Revenue'].sum().sort_values(ascending=False)
)

revenue_by_product.head(10)
```

```python
top_5_product = revenue_by_product.head(5)
top_5_product
```

```python
revenue_by_region = (
    df.groupby('Region')['Revenue'].sum().sort_values(ascending=False)
)

revenue_by_region.head(10)
```

```python
customer_type_revenue = (
    df.groupby('CustomerType')['Revenue'].sum().sort_values(ascending=False)
)

customer_type_revenue
```

```python
customer_type_summary = (
    df.groupby('CustomerType').agg(
        TotalRevenue=('Revenue', 'sum'),
        Orders=('OrderID', 'nunique'),
        AvgOrderValue=('Revenue', 'mean')
    ).sort_values('TotalRevenue', ascending=False)
)

customer_type_summary
```

```python
channel_summary = (
    df.groupby('SalesChannel')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

channel_summary
```

```python
return_summary = (
    df.groupby('Returned').agg(
        Orders=('OrderID', 'nunique'),
        Revenue=('Revenue', 'sum')
    )
)

return_summary
```

```python
df['Month'] = df['OrderDate'].dt.to_period('M')

monthly_revenue = (
    df.groupby('Month')['Revenue'].sum()
)

monthly_revenue.head()
```

```python
print('TOP 5 PRODUCTS')
print(top_5_product)

print('\nREVENUE BY REGION')
print(revenue_by_region)

print('\nCUSTOMER TYPE SUMMARY')
print(customer_type_summary)
```
